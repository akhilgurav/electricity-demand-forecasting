```python
from pyspark.sql import functions as F

base_path = "/Volumes/electricity_catalog/electricity_schema/raw/"

karnataka_weather = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(base_path + "Karnataka.csv")
)

maharashtra_weather = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(base_path + "Maharashtra.csv")
)

display(karnataka_weather.limit(10))
display(maharashtra_weather.limit(10))
```

```python
peak_raw = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(base_path + "corrected_peak_met_MW.csv")
)

peak_clean = peak_raw

for old_col in peak_clean.columns:
    new_col = (
        old_col
        .strip()
        .replace(" ", "_")
        .replace(",", "_")
        .replace(";", "_")
        .replace("{", "_")
        .replace("}", "_")
        .replace("(", "_")
        .replace(")", "_")
        .replace("\n", "_")
        .replace("\t", "_")
        .replace("=", "_")
    )
    
    peak_clean = peak_clean.withColumnRenamed(
        old_col,
        new_col
    )

peak_clean = peak_clean.withColumnRenamed(
    "Date",
    "date"
)
```

```python
energy_clean = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(base_path + "corrected_peak_met_MW.csv")
)

for old_col in energy_clean.columns:
    new_col = (
        old_col
        .strip()
        .replace(" ", "_")
        .replace(",", "_")
        .replace(";", "_")
        .replace("{", "_")
        .replace("}", "_")
        .replace("(", "_")
        .replace(")", "_")
        .replace("\n", "_")
        .replace("\t", "_")
        .replace("=", "_")
    )
    
    energy_clean = energy_clean.withColumnRenamed(
        old_col,
        new_col
    )

energy_clean = energy_clean.withColumnRenamed(
    "Date",
    "date"
)
```

```python
energy_clean.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "electricity_catalog.electricity_schema.bronze_energy"
    )
```

```python

from pyspark.sql import functions as F

energy = spark.table(
    "electricity_catalog.electricity_schema.bronze_energy"
)

value_cols = [c for c in energy.columns if c != "Date"]

# Count rows that are identical across every column
exact_duplicate_excess = (
    energy.count()
    - energy.dropDuplicates().count()
)

# Count dates that have conflicting records
date_stats = (
    energy.groupBy("Date")
    .agg(
        F.count("*").alias("rows"),
        F.countDistinct(
            F.struct(*[F.col(c) for c in value_cols])
        ).alias("distinct_records")
    )
)

date_stats = date_stats.filter(F.col("rows") > 1)

print("Exact duplicate excess rows:", exact_duplicate_excess)
print(
    "Dates with conflicting records:",
    date_stats.filter(F.col("distinct_records") > 1).count()
)
print(
    "Dates with only identical repeated records:",
    date_stats.filter(F.col("distinct_records") == 1).count()
)
```

```python

from pyspark.sql import functions as F

source_table = "electricity_catalog.electricity_schema.bronze_energy"

energy = spark.table(source_table)
value_cols = [c for c in energy.columns if c != "Date"]

# Remove only rows identical across every column
exact_clean = energy.dropDuplicates()

# Find dates with multiple distinct records
date_stats = exact_clean.groupBy("Date").agg(
    F.count("*").alias("row_count")
)

conflict_dates = (
    date_stats
    .filter(F.col("row_count") > 1)
    .select("Date")
)

# Preserve all conflicting records for investigation
conflicts = exact_clean.join(conflict_dates, "Date", "inner")

# Create a candidate table, retaining conflicting dates for now
exact_clean.write.format("delta").mode("overwrite").option(
    "overwriteSchema", "true"
).saveAsTable(
    "electricity_catalog.electricity_schema.bronze_energy_exact_clean"
)

conflicts.write.format("delta").mode("overwrite").option(
    "overwriteSchema", "true"
).saveAsTable(
    "electricity_catalog.electricity_schema.bronze_energy_conflicts"
)

print("Exact-clean candidate rows:", exact_clean.count())
print("Conflicting dates:", conflict_dates.count())
print("Conflicting records:", conflicts.count())
```

```python

from pyspark.sql import functions as F

energy = spark.table(
    "electricity_catalog.electricity_schema.bronze_energy_exact_clean"
)

conflicts = spark.table(
    "electricity_catalog.electricity_schema.bronze_energy_conflicts"
)

# Dates whose records conflict
conflict_dates = conflicts.select("Date").distinct()

# Keep dates with exactly one distinct record
modeling_energy = energy.join(
    conflict_dates, on="Date", how="left_anti"
)

modeling_energy.write.format("delta").mode("overwrite").option(
    "overwriteSchema", "true"
).saveAsTable(
    "electricity_catalog.electricity_schema.bronze_energy_modeling"
)

print("Modeling rows:", modeling_energy.count())
print("Distinct dates:", modeling_energy.select("Date").distinct().count())
print(
    "Duplicate dates remaining:",
    modeling_energy.groupBy("Date").count()
    .filter(F.col("count") > 1).count()
)
```
